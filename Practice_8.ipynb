Работа с API

```python
import requests
import pandas as pd
```

Получение информации обо всех книгах и сохранение ее в pd.DataFrame

```python
api_base_url = 'https://www.anapioficeandfire.com/api'
books_list = requests.get(
    f'{api_base_url}/books'
).json()

df_books = pd.DataFrame(books_list)
df_books.head()
```

Получение информации обо всех домах Вестероса (Houses) и сохранение ее в pd.DataFrame

```python
houses_list = requests.get(
    f'{api_base_url}/houses'
).json()

df_houses = pd.DataFrame(houses_list)
df_houses.head()
```

Получение информации обо всех домах Вестероса, у которых есть девиз, и сохранение ее в pd.DataFrame

```python
idx = df_houses[df_houses['words'] != '']
df_houses = pd.DataFrame(idx)
df_houses.head()
```
