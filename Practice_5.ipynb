Считывание датасета из файла Housing.csv

```python
import pandas as pd
import numpy as np
df_h = pd.read_csv('Housing.csv')
df_h.info()
```

Определение оптимального с точки зрения потребления памяти типа данных

```python
df_h.describe()
```

Во всех столбцах используются целочисленные значения. Минимальные значения во всех столбцах положительные, поэтому можно использовать следующий тип данных uint для столбцов:
price - uint32 (значения не превышают 4,2*10^9)
area - uint16 (значения не превышают 65535)
bedrooms - uint8 (значения не превышают 255)
bathrooms - uint8 (значения не превышают 255)
stories - uint8 (значения не превышают 255)
parking - uint8 (значения не превышают 255)

Изменение типа данных столбцов датафрейма на новый и сравнение потребление памяти до и после оптимизации

```python
df_h = df_h.fillna(0)
df_h['price'] = df_h['price'].astype(np.uint32)
df_h['area'] = df_h['area'].astype(np.uint16)
df_h['bedrooms'] = df_h['bedrooms'].astype(np.uint8)
df_h['bathrooms'] = df_h['bathrooms'].astype(np.uint8)
df_h['stories'] = df_h['stories'].astype(np.uint8)
df_h['parking'] = df_h['parking'].astype(np.uint8)
df_h.info()
```

Потребление памяти до оптимизации было 55.5 KB, после оптимизации стало 35.3 KB. Используемый объем потребляемой памяти уменьшился на 36 %.
