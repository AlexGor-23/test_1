Вывод самого популярного мужского и самого популярного женского имени на корабле

```python
def first_name_male(name):
    first = name.str.extract(r'Mr\.\s+(\w+)', expand=False)
    first.loc[first.isna()] = name.str.extract(r'\.\s+(\w+)', expand=False)
    return first
male_names = first_name_male(df.loc[df['Sex']=='male','Name'])
print(f'Самое популярное мужское имя {male_names.value_counts().index[0]} встречается {male_names.value_counts().iloc[0]} раз')
def first_name_female(name):
    first = name.str.extract(r"Mrs\.\s+[^(]*\((\w+)", expand=False)
    first.loc[first.isna()] = name.str.extract(r"\.\s+(\w+)", expand=False)
    return first
female_names = first_name_female(df.loc[df['Sex']=='female','Name'])
print(f'Самое популярное женское имя {female_names.value_counts().index[0]} встречается {female_names.value_counts().iloc[0]} раз')
```

Вывод самого популярного мужского и самого популярного женского имени на корабле в каждом классе

```python
df = pd.read_csv('train.csv')
df['Name'] = df['Name'].apply(lambda x: x[x.find('.')+2:])
df['Name'] = df['Name'].apply(lambda x: x[x.find('(')+1:])
df['Name'] = df['Name'].apply(lambda x : x if x.find(' ') == -1 else x[:x.find(' ')])
df.groupby("Pclass")["Name"].value_counts()

for i in range(3):
  print(f"Самое популярное мужское имя {i+1} класса - {df[(df['Sex'] == 'male') & (df['Pclass'] == i+1)]['Name'].value_counts().index[0]} - встречается {df[(df['Sex'] == 'male') & (df['Pclass'] == i+1)]['Name'].value_counts().iloc[0]} раз")
  print(f"Самое популярное женское имя {i+1} класса - {df[(df['Sex'] == 'female') & (df['Pclass'] == i+1)]['Name'].value_counts().index[0]} - встречается {df[(df['Sex'] == 'female') & (df['Pclass'] == i+1)]['Name'].value_counts().iloc[0]} раз")

```

Вывод количества n-местных кабин (в которых было 2, 3, 4, ... человека)
