```python
class Chocolate:
  def __init__(self, price:float, additive=None):
    self.price = price
    self.additive = additive
    if price < 0:
      raise ValueError('Указана отрицательная цена')
  def show_my_chocolate(self):
    # return f'Шоколадный батончик, {self.additive}'
    if self.additive == None:
      print('Обычный шоколадный батончик')
    else:
      print(f'Шоколадный батончик, добавка: {self.additive}')
    #return f'Шоколадный батончик, добавка: {self.additive}'
  def __str__(self):
    return f'Шоколадный батончик, {self.price} рублей'
```

```python
chocolate = Chocolate(100,'jam')
chocolate.show_my_chocolate()
print(chocolate)
```

```python
chocolate = Chocolate(10)
chocolate.show_my_chocolate()
print(chocolate)
```

```python
chocolate = Chocolate(-10)
chocolate.show_my_chocolate()
print(chocolate)
```
